# Demographics × Course Preference Analysis

## Objective
Analyze how learner demographics (Age and Gender) influence course preferences and enrollment patterns.

### Analysis Performed
- Created age groups for learners.
- Compared age groups with course categories.
- Compared gender with course levels.
- Identified learner segments and their preferences.

### Goal
Understand which demographic groups prefer different types of courses.

```python
# Create Age Groups
df["AgeGroup"] = pd.cut(
    df["Age_x"],
    bins=[15, 20, 25, 30, 35],
    labels=["16-20", "21-25", "26-30", "31-35"]
)

df["AgeGroup"].value_counts()
```

```python
plt.figure(figsize=(12,6))

heatmap_data = pd.crosstab(df["AgeGroup"], df["CourseCategory"])

sns.heatmap(heatmap_data, annot=True, cmap="YlGnBu", fmt="d")

plt.title("Age Group vs Course Category")
plt.xlabel("Course Category")
plt.ylabel("Age Group")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="CourseLevel",
    hue="Gender_x",
    palette="Set2"
)

plt.title("Gender vs Course Level")
plt.show()
```

```python
gender_course = pd.crosstab(df["Gender_x"], df["CourseLevel"])

print(gender_course)
```

# Behavioral Insights

## Objective
Analyze learner behavior to understand engagement, course completion patterns, and learning preferences.

### Analysis Performed
- Calculated the average number of courses taken per learner.
- Identified the most active learners.
- Compared beginner and advanced learner behavior.
- Analyzed enrollment concentration among learners.

### Goal
Identify behavioral trends that can help improve learner engagement and course recommendations.

```python
avg_courses = df.groupby("UserID")["CourseID"].nunique().mean()

print("Average Courses Taken Per Learner:", round(avg_courses, 2))
```

```python
active_learners = (
    df.groupby("UserID")["CourseID"]
    .count()
    .sort_values(ascending=False)
    .head(10)
)

print(active_learners)
```

```python
plt.figure(figsize=(10,5))

active_learners.plot(kind="bar", color="skyblue")

plt.title("Top 10 Most Active Learners")
plt.xlabel("User ID")
plt.ylabel("Number of Enrollments")
plt.xticks(rotation=45)

plt.show()
```

```python
course_level = df["CourseLevel"].value_counts()

print(course_level)
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="CourseLevel", palette="pastel")

plt.title("Beginner vs Intermediate vs Advanced Courses")

plt.show()
```

```python
enrollment = df.groupby("UserID")["CourseID"].count()

print(enrollment.describe())
```

```python
plt.figure(figsize=(8,5))

plt.hist(enrollment, bins=15)

plt.title("Enrollment Concentration Among Learners")
plt.xlabel("Courses Enrolled")
plt.ylabel("Number of Learners")

plt.show()
```
